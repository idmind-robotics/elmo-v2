[POST] http://<robot_ip>:8001/command 
 - Send a command to the robot

Body:

```python
{
    "op": <operation>
    "<param1>": <value1> 
}
```

Reply:

```python
{
    "success": <true or false>,
    "message": <OK or error description>
}
```

Choose the operation by passing the appropriate value for `op` and any additional parameters required.

As an example, to disable the `blush` behaviour, which is the robot's reaction to pets, the following body woud be sent to the `/command` endpoint:

```python
{
    "op": "enable_behaviour",
    "name": "blush",
    "control": false
}
```

> #### Full list of available operations:

 - ### enable_behaviour

<u>Description</u>: control behaviours

<u>Expected params</u>:

    - <string> name
    - <bool> control

- ### set_pan_torque

<u>Description</u>: control pan torque

<u>Expected params</u>:

    - <bool> control

- ### set_pan

<u>Description</u>: control pan angle

<u>Expected params</u>:

    - <float> angle

- ### set_tilt_torque

<u>Description</u>: control tilt torque

<u>Expected params</u>:

    - <bool> control

- ### set_tilt

<u>Description</u>: control tilt angle

<u>Expected params</u>:

    - <float> angle

- ### play_sound

<u>Description</u>: play a sound from the robot's internal multimedia server

<u>Expected params</u>:

    - <string> name

- ### pause_audio

<u>Description</u>: Pause sound playback

<u>Expected params</u>:

- ### set_volume

<u>Description</u>: control volume

<u>Expected params</u>:

    - <int> volume (0 ~ 100)

- ### start_recording

<u>Description</u>: start recording audio. recording will be saved as mic.wav

<u>Expected params</u>:

- ### stop_recording

<u>Description</u>: stop recording audio

<u>Expected params</u>:

- ### set_screen

<u>Description</u>: update screen. Use this to set an image, play a video, show text, or load a website. send an empty object to clear.

<u>Expected params</u>:

    - <string> image (optional)
    - <string> video (optional)

<u>Description</u>: update led matrix. send a list of 13x13=169 colors. each color is a list of 3 (rgb components). row major ordering.

<u>Expected params</u>:

<u>Description</u>: load an icon to the led matrix using the robot's internal media server

<u>Expected params</u>:
