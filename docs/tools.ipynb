# **Recommended Tools**

> ### UV - For virtual Environments

### Useful commands:

- Create a clean virtual environment ::: **uv venv**
- Create a clean virtual environment in a specific Python version ::: **uv venv --python x.xx.xx**
- Activate the environment ::: **source .venv/bin/activate**
- Sync dependencies from the project requirements ::: **uv pip install -r requirements.txt**
- View available Python versions ::: **uv python list**
- Unninstall a version of Python ::: **uv python unninstall**
- Run a specific script ::: **uv run**
- Install libraries on the environment ::: **uv pip install**
- List all installed libraries in the environment ::: **uv pip list**

###### Note: For more information about uv, check out their [Getting Started Guide](https://docs.astral.sh/uv/getting-started/).
