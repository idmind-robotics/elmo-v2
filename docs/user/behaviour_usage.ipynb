# **Behaviour Usage**

This is the documentation on how to use the behaviours and how to trigger them.

On most of the behaviours, it will be asked for you to select different modes. To do so, touch the power button in the back, and see the LEDs change on the chest - that's the mode selector.

> ### **DEV Version:**

#### `Blush`:

**Step 1** ::: Select the Idle mode (default/startup mode).

 

**Step 2** ::: Touch TAGI's head.

- Side effect ::: TAGI will fall in love with you.

#### `Clock`:

**Step 1** ::: Select the Idle mode (default/startup mode).

 

**Step 2** ::: Touch TAGI's chest.

- Side effect ::: Shows time and weather on the chest.

#### `Hello`: 

**Step 1** ::: Select the Idle mode (default/startup mode).

 

**Step 2** ::: Show up in front of TAGI's head.

- Side effect ::: TAGI will say hello.

#### `Wifi Connect`:

**Step 1** ::: Select Wifi mode. 



**Step 2** ::: TAGI's face will show what it is seeing. Show it the QR Code.

- Side effect ::: TAGI will connect to the Wifi provided.

#### `Look Around`: 

**Step 1** Turn on the robot.

- Side effect ::: Head will move around.

#### `Test Motors`: 

**Step 1** Turn on the robot.

- Side effect ::: Head will move around.

> ### **Kiosk Version:**

#### `Photographer`:

**Step 1** ::: Select the Photographer mode.



**Step 2** ::: Prepare to take a picture. Stay in the position you want to be in

- Side effect ::: TAGI will center the face, initiate a 10 second countdown on the chest and print the picture.

#### `Finger Counter`:

**Step 1** ::: Select the Idle mode (default/startup mode).



**Step 2** ::: Show your hand to TAGI and hold a number of fingers between 1 and 5.

- Side effect ::: TAGI will show the fingers lifted on the screen.

#### `Conversation (using various LLMs)`:

**Step 1** ::: Select the Conversation mode.



**Step 2** ::: Talk to TAGI.

- Side effect ::: TAGI will answer you.
