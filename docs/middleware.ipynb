# **Middleware Documentation**

The Middleware documentation should be considered for code development, by using the several classes which extend the DBEntry class.

This library acts as a **Redis-based** Middleware designed for TAGI. It facilitates inter-process communication, state management, and hardware abstraction by using Redis as a centralized, JSON-serialized data store.

For further information about Redis, check the their [official website](https://redis.io/docs/latest/).

> ## **Library Architecture**
The library is built on three pillars:

1. **Global Key-Value Store:** Low-level functions to interface with Redis using JSON serialization.
2. **Node Management:** Logic to track, monitor, and control distributed processes (Nodes).
3. **Data Models (DBEntry):** An ORM-like abstraction that maps Python class properties directly to Redis keys.

> ##### 1. Global Redis Utilities

These functions handle the direct communication with the Redis instance.

Functions
- ``get_connection()`` ::: Returns a standard redis.Redis() connection object
- ``set_key(key,value)`` ::: Serializes value to JSON and stores it in Redis.
- ``get_key(key)`` ::: Retrieves a key and deserializes the JSON content.
- ``has_key(key)`` ::: Returns True if the specific key exists.
- ``delete_all()`` ::: Flushes the entire Redis database.

> ##### 2. Process & Node Management

This section manages the lifecycle of different software modules (Nodes) running on the robot.

**``Node`` Class**

Represents a single running process. Upon initialization, it registers its PID in Redis.

- **Logging:** Supports ``loginfo``, ``logwarn``, and ``logerror`` with configurable levels.
- **Shutdown:** Nodes can check ``is_shutdown()`` to perform a graceful exit.

**``NodeManager`` Class**

A tool for orchestrating multiple nodes.

- ``list_nodes()`` ::: Scans Redis for all active node names.
- ``is_alive(name)`` ::: Verifies if a node is registered in Redis and its process is still running in the OS.
- ``force_shutdown(name)`` ::: Sends a SIGKILL to the process and cleans up Redis keys.



> #### 3. Hardware & State Models

#### **Classes:**

**Robot**
- Responsability ::: General metadata
- Key Fields ::: ``name``

**Camera**
- Responsability ::: Vision state
- Key Fields ::: ``url``, ``take_picture``, ``face_detected``

**LEDS**
- Responsability ::: LED Matrix (13x13)
- Key Fields ::: ``colors``, ``brightness``, ``load_from_url()``

**Battery**
- Responsability ::: Power monitoring
- Key Fields ::: ``voltage``, ``percentage``, ``i2c_address``

**TouchSensors**
- Responsability ::: Input detection
- Key Fields ::: ``touch_chest``, ``touch_head_0..4``

**Pan / Tilt**
- Responsability ::: Motor control
- Key Fields ::: ``angle``, ``pid_p``, ``temperature``

**Speech**
- Responsability ::: TTS control
- Key Fields ::: ``say``, ``language``, ``saying``

**Server**
- Responsability ::: Asset management
- Key Fields ::: Generates URLs for images, sounds, and videos.
